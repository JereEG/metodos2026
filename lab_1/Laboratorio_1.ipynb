
## Ejercicio 1: Suma de N números positivos
### Objetivo: Sumar N números positivos y medir tiempos para un N entre 0 y 255 y para un N = 10.000. 
### Para ello se diaseño el modelo matematico y el algoritmo en pseudocodigo y se  clasifico como Iterativo y directo


### a) Modelo Matemático
$N \in \mathbb{N}$:
* **Formula matematica:**
  $$S = \sum_{i=1}^{N} X_i \quad \text{con } X_i > 0 \quad \forall i \in \{1, 2, \dots, N\}$$

---

### b) Algoritmo en Pseudocódigo y Clasificación
* **Clasificación:** Algoritmo **Directo** (alcanza el resultado exacto en un número finito y predeterminado de $N$ operaciones).

```text
ALGORITMO SumarNnumeros
VAR
    ENTERO: N, i
    REAL: numero, total
INICIO
    total = 0;
    numero = 0;
    ESCRIBIR "Ingrese cuantos numeros quiere sumar N: ";
    LEER N;
    SI (N <= 0) ENTONCES
        ESCRIBIR "Error: la cantidad de numeros a sumar debe ser mayor a 0 ";
    SINO
        PARA i = 1 HASTA N HACER
            ESCRIBIR "Ingrese el numero a sumar: "
            LEER numero
            MIENTRAS (numero <= 0) HACER
                ESCRIBIR "Error: Solo se pueden sumar valores positivos, digite de nuevo: ";
                LEER numero;
            FIN-MIENTRAS;
            total = total + numero;
        FIN-PARA;
        ESCRIBIR "La suma total es: ", total;
    FIN-SI;
FIN.

````python
import random
import time
import shutil

#Esta funcion me permite seleccioanar que rango de prueba quiere sea n=[0,250] o n=10.000
def solicitar_cantidad_n() -> int:
    print("\n--- Seleccione el rango de prueba para N ---")
    print("[1] Probar con un rango corto de [0 - 255]")
    print("[2] Probar un rango amplio (hasta 10.000)")
    
    while True:
        opcionion = input("Ingrese opción [1] o [2]: ").strip()
        if opcion == "1":
            limite = 255
            break
        elif opcion == "2":
            limite = 10000
            break
        print("Opción inválida. Ingrese una de las dos opciones [1] o [2]")

    while True:
        try:
            n = int(input(f"Ingrese el valor de N o cantidad de numeros que quiere sumar en este rango [mínimo 1, máximo {limite}]: "))
            if 1 <= n <= limite:
                return n
            
            print(f"Error: N debe estar en el intervalo [1, {limite}].")
        except ValueError:
            print("Error: Debe ingresar un número entero válido.")
        
 #Con esta funcion puedo elegir entre ingresar los N números manualmente o generarlos al azar con random
def obtener_datos_entrada(n: int) -> list[float]:
    numeros = []
    print(f"\n--- Elija como quiere cargar los {n} numeros a suamr ---")
    print("1) Generar valores aleatorios [random] positivos automáticamente")
    print("2) Ingresar valores manualmente por teclado")
    
    while True:
        modo = input("Seleccione la forma de cargar los numeros [1] o [2] : ").strip()
        if modo in ("1", "2"):
            break
        print("Opción inválida. Seleccione [1] o [2].")
       
    if modo == "1":
        # Genera N valores de punto flotante positivos entre 0.10 y 100.00 utilizando la libreria random
        numeros= [round(random.uniform(0.1, 100.0), 4) for i in range(n)]
        return numeros
    else:
        print(f"\nIngrese a continuación los {n} números positivos:")
        for i in range(n):
            while True:
                try:
                    val = float(input(f"  Número [{i + 1}/{n}] (> 0): "))
                    if val > 0:
                        numeros.append(val)
                        break
                    print("  Error: El número debe ser estrictamente positivo (x > 0).")
                except ValueError:
                    print("  Error: Ingrese un valor numérico válido.")
        return numeros
    
#  Muestra que valores que componen la sumatoria para verificarlo que mi programa cumple el objetivo cuando uso random sobre todo para los numeros
def mostrar_sumandos(numeros: list[float]):
    total = len(numeros)
    print(f"\n--- Listado completo de números a sumar con N={total} ---")

    # Formatea cada número con 4 decimales alineados
    elementos = [f"{x:8.4f}" for x in numeros]

    # Detecta el ancho disponible en la terminal del usuario (por defecto 80 columnas)
    ancho_consola = shutil.get_terminal_size(fallback=(80, 24)).columns
    ancho_item = 10  # 8 caracteres del número + 2 de separación ", "
    columnas = max(1, ancho_consola // ancho_item)

    # Imprime en bloques que llenan el ancho de la pantalla de corrido
    for i in range(0, total, columnas):
        fila = elementos[i:i + columnas]
        print(", ".join(fila))

    print("--- Fin del listado ---")
    
#Calcula la suma de una lista de números positivos y mide tiempos.
def sumar_n_positivos(numeros: list[float]) -> tuple[float, float, float]:
    #calculamos los tiempos con la libreria time para mostrar los tiempo de ejecucion y comparar
    tiem_clock_icio_clock = time.perf_counter()
    tiem_clock_icio_cpu = time.process_time()
    
    total = 0.0
    for x in numeros:
        if x <= 0:
            raise ValueError(f"Valor no positivo detectado: {x}")
        total += x
        
    t_fin_cpu = time.process_time()
    t_fin_clock = time.perf_counter()
    
    tiem_reloj_f = t_fin_clock - tiem_clock_icio_clock
    tiem_cpu_f = t_fin_cpu - tiem_clock_icio_cpu
    
    return total, tiem_reloj_f, tiem_cpu_f


def main():
    #1. solicitar el N con el que trabajo
    n = solicitar_cantidad_n()
    
    #2. generar o obtener lso numeros a sumar
    numeros = obtener_datos_entrada(n)
    
    #3. Visualización de los sumandos
    mostrar_sumandos(numeros)

    #4. Procesamiento y cálculo de la suma
    total, t_clock, t_cpu = sumar_n_positivos(numeros)
    
    #5. Presentación de resultados analíticos
    print("=" * 65)
    print("                    RESULTADOS DE LA EJECUCION DEL EJERCICIO")
    print("=" * 65)
    print(f"Cantidad de elementos sumados (N) : {n}")
    print(f"Suma total calculada              : {total:,.4f}")
    print(f"Tiempo transcurrido (Reloj)       : {t_clock:.8f} segundos")
    print(f"Tiempo de uso de CPU              : {t_cpu:.8f} segundos")
    print("=" * 65)

if __name__ == "__main__":
    main()
````

Ejercicio 2: Suma de Valores Pares de 200 Números Positivos

a) Modelo Matemático

Variables de entrada:Conjunto de 200 números  positivos $\{X_1, X_2, \dots, X_{200}\} \subset \mathbb{Z}^+$, tal que $X_i > 0, \; \forall i \in \{1, 2, \dots, 200\}$.

Variable de salida:$S \in \mathbb{Z}^+$: suma acumulada de los elementos pares.
$$S = \sum_{i=1}^{200} f(X_i)$$donde la función  $f(X_i)$ se define como:$$f(X_i) = \begin{cases} X_i & \text{si } X_i \bmod 2 = 0 \\ 0 & \text{si } X_i \bmod 2 \neq 0 \end{cases}$$

**b) Algoritmo en Pseudocódigo y Clasificación**

Clasificación: Algoritmo **Directo** ya que alcanza el resultado exacto en un número finito y predeterminado de $N$ operaciones o pasos

```text
Algoritmo en Pseudocódigo y Clasificación
ALGORITMO SumaPares
VAR
    ENTERO: i;
    REAL: numero, total;
INICIO
    total = 0;
    PARA i = 1 HASTA 200 HACER
        ESCRIBIR "Ingrese el numero positivo ", i, ":";
        LEER numero;
        SI (numero MOD 2 == 0) ENTONCES
            total = total + numero;
        FIN-SI;
    FIN-PARA;
    ESCRIBIR "La suma de los números pares es: ", total;
FIN.

````python
import random
import time

def sumar_pares(numeros: list[int]) -> tuple[int, int, float, float]:
    tiem_reloj_i, tiem_cpu_i = time.perf_counter(), time.process_time()
    
    suma_pares = 0
    cant_pares = 0
    for x in numeros:
        if x % 2 == 0:
            suma_pares += x
            cant_pares += 1
            
    tiem_reloj_f= time.perf_counter() - tiem_reloj_i
    tiem_cpu_f = time.process_time() - tiem_cpu_i
    return suma_pares, cant_pares, tiem_reloj_f, tiem_cpu_f

def mostrar_sumandos(numeros: list[int]):
    print(f"\n--- Listado de los {len(numeros)} números a sumar ---")
    
    for i in range(0, len(numeros), 10):
        bloque = numeros[i : i + 10]
        linea = "".join(f"{x:6d}" for x in bloque)
        print(linea)

    print("--- Fin del listado ---")

if __name__ == "__main__":
    N = 200
    print("=== EJERCICIO 2: SUMA DE PARES DE 200 NÚMEROS POSITIVOS ===")
    opcion = input("Seleccione modo: [1] Generar 200 aleatorios | [2] Ingreso manual: ").strip()

    if opcion == "2":
        print(f"\nIngrese los {N} números enteros positivos:")
        numeros = [int(input(f"Número {i+1}: ")) for i in range(N)]
    else:
        random.seed(42)
        numeros = [random.randint(1, 500) for _ in range(N)]

    total, pares, tiem_cpu_f, tiem_reloj_f = sumar_pares(numeros)

    print("\n--- RESULTADOS ---")
    print(f"Total de datos procesados (N) : {len(numeros)}")
    mostrar_sumandos(numeros)
    print(f"\nCantidad de pares sumados     : {pares}")
    print(f"Suma total de pares           : {total}")
    print(f"Tiempo de ejecución (Reloj)   : {tiem_reloj_f:.8f} s")
    print(f"Tiempo de CPU                 : {tiem_cpu_f:.8f} s")
````

EJERCICIO 3

Obtener los $N$ primeros números de la sucesión de Fibonacci. Probar con $N = 200$

a) Formulación del Modelo Matemático

Variables de entrada:$N \in \mathbb{N}$: cantidad total de términos a generar de la sucesión ($N \ge 1$).

Variable de salida:Secuencia finita y ordenada $\{F_1, F_2, \dots, F_N\} \subset \mathbb{N}_0$.

Formula o modelo matematico:

$$F_n = \begin{cases} 0 & \text{si } n = 1 \\ 1 & \text{si } n = 2 \\ F_{n-1} + F_{n-2} & \text{si } n \ge 3 \end{cases} \quad \text{con } n \in \mathbb{N}$$

**b) Algoritmo en Pseudocódigo y Clasificación**

Clasificación: algoritmo Directo e Iterativo ya que alcanza el resultado en un numero finito y determinado  de $N$ operaciones elementales conocidas a priori utilizando un ciclo finito para resolver el problema

```text
ALGORITMO Fibonacci.
VAR
    ENTERO: N, i,a, b, c
INICIO
    ESCRIBIR "Ingrese la cantidad de terminos N: ";
    LEER N;
    
    SI (N <= 0) ENTONCES
        ESCRIBIR "Error: El valor de N debe ser mayor a 0.";
    SINO
        a = 0;
        b = 1;
        
        PARA i = 1 HASTA N HACER
            SI (i == 1) ENTONCES
                ESCRIBIR a;
            SINO
                SI (i == 2) ENTONCES
                    ESCRIBIR b;
                SINO
                    c = a + b;
                    ESCRIBIR c;
                    a = b;
                    b = c;
                FIN-SI.
            FIN-SI.
        FIN-PARA.
    FIN-SI.
FIN.

````python
def fibonacci(n):
    """
    Genera los n primeros números de la sucesión de Fibonacci.
    """
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    elif n == 2:
        return [0, 1]

    fib = [0, 1]
    for _ in range(2, n):
        fib.append(fib[-1] + fib[-2])
    return fib


if __name__ == "__main__":
    try:
        N = int(input("Ingrese la cantidad de términos de Fibonacci a generar: "))
        fib_n = fibonacci(N)
        print(f"Los {N} primeros números de Fibonacci son:")
        print(fib_n)

        print(f"Cantidad generada: {len(fib_n)}")
    except ValueError:
        print("Por favor, ingrese un número entero válido.")
````

Ejercicio 4a) Formulación del Modelo Matemático

Traducción del enunciado: Al quíntuple de un número ($5x$) se le suma el triple de su cuadrado ($3x^2$), obteniéndose el triple del mismo ($3x$) más 1:
$$5x + 3x^2 = 3x + 1$$Forma canónica de la ecuación cuadrática ($ax^2 + bx + c = 0$):$$3x^2 + 5x - 3x - 1 = 0 \implies 3x^2 + 2x - 1 = 0$$Donde los coeficientes son:$$a = 3, \quad b = 2, \quad c = -1$$Aplicando la fórmula resolvente:$$x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a} = \frac{-2 \pm \sqrt{2^2 - 4(3)(-1)}}{2(3)} = \frac{-2 \pm \sqrt{4 + 12}}{6} = \frac{-2 \pm 4}{6}$$Obtenemos las raíces reales: $x_1 = \frac{1}{3} \approx 0,333333$ y $x_2 = \frac{-6}{6} = -1$

b)Clasificación del algoritmo: Directo ya que resuelve el problema en una cantidad finita, fija y predeterminada de pasos
```text
ALGORITMO ObtenerRaices
VAR
    REAL: a, b, c, disc, x1, x2
INICIO
    a = 3;
    b = 2;
    c = -1;
    x1 = 0;
    x2 = 0;
    disc = 0;

    disc = (b * b) - (4 * a * c);

    SI (disc < 0) ENTONCES
        ESCRIBIR "No posee raíces reales.";
    SINO
        SI (disc == 0) ENTONCES
            x1 = -b / (2 * a);
            ESCRIBIR "Posee una raíz única: ", x1;
        SINO
            x1 = (-b + RC(disc)) / (2 * a);
            x2 = (-b - RC(disc)) / (2 * a);
            ESCRIBIR "Posee dos raíces reales: ", x1, " y ", x2;
        FIN-SI;
    FIN-SI;
FIN.

````python
import math
import time

def resolver_cuadratica(a, b, c):
    if a == 0:
        print("El coeficiente 'a' no puede ser 0 en una ecuación cuadrática.")
        return None

    t_ini = time.perf_counter()
    disc = b**2 - 4*a*c

    if disc < 0:
        res = "Sin raíces reales"
    elif disc == 0:
        x = -b / (2*a)
        res = (x,)
    else:
        x1 = (-b + math.sqrt(disc)) / (2*a)
        x2 = (-b - math.sqrt(disc)) / (2*a)
        res = (x1, x2)

    duracion = time.perf_counter() - t_ini
    return res, disc, duracion

if __name__ == "__main__":
    print("=== EJERCICIO 4: ECUACIÓN CUADRÁTICA ===")
    print("Ecuación planteada: 3x^2 + 2x - 1 = 0")
    
    # Coeficientes obtenidos del modelo
    a, b, c = 3.0, 2.0, -1.0
    
    resultado, discriminante, tiempo = resolver_cuadratica(a, b, c)

    print("\n--- RESULTADOS ---")
    print(f"Discriminante (b^2 - 4ac) : {discriminante:.4f}")
    if isinstance(resultado, tuple):
        if len(resultado) == 2:
            print(f"Raíz x1                  : {resultado[0]:.6f} (1/3)")
            print(f"Raíz x2                  : {resultado[1]:.6f} (-1)")
        else:
            print(f"Raíz única               : {resultado[0]:.6f}")
    else:
        print(resultado)
        
    print(f"Tiempo de ejecución      : {tiempo:.8f} segundos")
````

a) Formulación del Modelo Matemático
$$S_N = \sum_{i=0}^{N} \frac{1}{i!} = 1 + \frac{1}{1!} + \frac{1}{2!} + \frac{1}{3!} + \dots + \frac{1}{N!} \quad \text{donde } 0! = 1$$

Clasificación: Iterativo y convergente. La aproximación mejora al aumentar la cantidad de sumandos.

```text
ALGORITMO Numero_Euler.
VAR
    ENTERO: N, i
    REAL: e, factorial
INICIO
    ESCRIBIR "Ingrese la cantidad de terminos N: ";
    LEER N;

    SI (N <= 0) ENTONCES
        ESCRIBIR "Error: N debe ser mayor a 0.";
    SINO
        e = 1.0;
        factorial = 1.0;

        PARA i = 1 HASTA N HACER
            factorial = factorial * i;
            e = e + (1.0 / factorial);
        FIN-PARA;

        ESCRIBIR "El valor aproximado de e es: ", e;
    FIN-SI;
FIN.

````python
from decimal import Decimal, getcontext
import math

def calcular_e_serie(N):
    """
    Calcula e usando la serie: e = 1 + 1/1! + 1/2! + 1/3! + ... + 1/N!
    """
    suma = 1.0  # Empezamos con el primer término (1)
    factorialorial = 1.0
    
    print("Aproximación de e usando la serie:")
    print(f"Término 0: e ≈ {suma:.10f}")
    
    for n in range(1, N + 1):
        factorialorial *= n  # Calculamos n! incrementalmente
        termino = 1.0 / factorialorial
        suma += termino
        
        print(f"Término {n}: e ≈ {suma:.10f} (1/{n}! = {termino:.10f})")
    
    return suma

def calcular_e_decimal(N, precision=50):
    """
    Calcula e usando Decimal para mayor precisión
    """
    getcontext().prec = precision
    
    suma = Decimal(1)
    factorialorial = Decimal(1)
    
    print(f"\nAproximación con Decimal (precisión {precision}):")
    print(f"Término 0: e ≈ {suma}")
    
    for n in range(1, N + 1):
        factorialorial *= Decimal(n)
        termino = Decimal(1) / factorialorial
        suma += termino
        
        print(f"Término {n}: e ≈ {suma}")
    
    return suma

def analizar_convergencia():
    """
    Muestra cuántos términos se necesitan para una buena aproximación
    """
    print("\n" + "="*50)
    print("ANÁLISIS DE CONVERGENCIA")
    print("="*50)
    
    valor_real = math.e
    suma = 1.0
    factorialorial = 1.0
    
    for n in range(1, 20):
        factorialorial *= n
        termino = 1.0 / factorialorial
        suma += termino
        error = abs(valor_real - suma)
        
        print(f"n={n:2d} | e ≈ {suma:.15f} | Error: {error:.2e}")
        
        if error < 1e-10:
            print(f"¡Convergencia alcanzada! Error < 1e-10 con n={n}")
            break

if __name__ == "__main__":
    print("CÁLCULO DEL NÚMERO e USANDO SERIE INFINITA")
    print("e = 1 + 1/1! + 1/2! + 1/3! + ...")
    print(f"Valor real de math.e: {math.e}\n")
    
    try:
        N = int(input("Ingrese el número de términos de la serie (recomendado: 15): "))
        
        # Calcular con float
        aprox_float = calcular_e_serie(N)
        
        # Calcular con Decimal
        aprox_decimal = calcular_e_decimal(N)
        
        # Resultados finales
        print("\n" + "="*60)
        print("RESULTADOS FINALES:")
        print("="*60)
        print(f"Valor real de math.e:      {math.e}")
        print(f"Aproximación con {N} términos (float):  {aprox_float:.15f}")
        print(f"Aproximación con {N} términos (Decimal): {aprox_decimal}")
        print(f"Diferencia con valor real:  {abs(math.e - aprox_float):.2e}")
        
        # Análisis de convergencia
        analizar_convergencia()
        
    except ValueError:
        print("Error: Por favor ingrese un número entero válido.")
    except Exception as e:
        print(f"Error inesperado: {e}")
````

````python
"""
MÉTODOS COMPUTACIONALES - LABORATORIO 1
Ejercicio 6: Cálculo de Pi mediante la serie de Gregory-Leibniz / Taylor.
Evaluación automática de términos y tiempos para 3, 4 y 5 dígitos decimales exactos.
"""
import math
import time

def calcular_pi_tolerancia(decimales: int):
    """
    Calcula Pi acumulando término a término hasta que el error absoluto
    sea menor a la tolerancia del criterio de Scarborough: tol = 0.5 * 10^(-decimales).
    """
    tol = 0.5 * (10 ** (-decimales))
    pi_ref = math.pi

    pi_aprox = 0.0
    signo = 1.0
    k = 0
    error = 1.0

    t_ini = time.perf_counter()

    while error >= tol:
        termino = signo * (4.0 / (2 * k + 1))
        pi_aprox += termino
        signo = -signo
        k += 1
        error = abs(pi_ref - pi_aprox)

    tiempo = time.perf_counter() - t_ini
    return k, pi_aprox, tol, error, tiempo

def main():
    print("=" * 80)
    print("  EJERCICIO 6: CÁLCULO DE PI POR SERIE Y ANÁLISIS DE CONVERGENCIA")
    print("=" * 80)
    print(f"Valor real de referencia (math.pi) : {math.pi:.15f}\n")

   
    casos = [3, 4, 5]

    print(f"{'Cifras (n)':<12}{'Tolerancia':<14}{'Términos (N)':<15}{'Pi aproximado':<18}{'Tiempo (s)':<12}")
    print("-" * 80)

    for n in casos:
        k, aprox, tol, err, t_ejec = calcular_pi_tolerancia(n)
        print(f"{n:<12}{tol:<14.6f}{k:<15,}{aprox:<18.7f}{t_ejec:<12.6f}")

    print("-" * 80)

if __name__ == "__main__":
    main()
````
